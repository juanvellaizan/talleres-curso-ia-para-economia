### **Ejercicio 1: Carga y Preparación Inicial de los Datos**

Primero, carguemos las librerías necesarias y el dataset. Nos enfocaremos en un subconjunto de las variables numéricas para mantener el taller manejable, pero el principio se aplica a todo el dataset.

```python
# Importa las librerías necesarias
### TU CÓDIGO AQUÍ ###
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Mejorar visualización de dataframes y gráficos

```python
# Que muestre todas las columnas
pd.options.display.max_columns = None
# En los dataframes, mostrar los float con dos decimales
pd.options.display.float_format = '{:,.2f}'.format

# Configuraciones para una mejor visualización
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

Cargar el dataset

```python
# Usamos el dataset Ames Housing desde su fuente original (requiere internet)
url = 'http://jse.amstat.org/v19n3/decock/AmesHousing.txt'
df = pd.read_csv(url, sep='\t')
df.head()
```

```python
# Para este taller, solo usaremos algunas columnas clave para simplificar.
df_sub = df[['Overall Qual', 'Gr Liv Area', 'SalePrice']].copy()
df_sub.columns = ['OverallQual', 'GrLivArea', 'SalePrice'] # Renombrar para facilidad
df_sub.head()
```

### **Ejercicio 2: Dividir el conjunto de datos**

El método `.info()` muestra que no hay nulos en nuestro subconjunto. ¡Perfecto! Ahora, definamos nuestras variables `X` (predictoras) e `y` (objetivo) y dividamos los datos.

```python
# Definir X e y
### TU CÓDIGO AQUÍ ###
# Definir X e y
X = df_sub[['OverallQual', 'GrLivArea']]
y = df_sub['SalePrice']
```

```python
# Dividir en entrenamiento y prueba (70/30)
### TU CÓDIGO AQUÍ ###
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

### **Ejercicio 3: Modelo con Sobreajuste**

Vamos a crear un modelo muy complejo para ver qué tan mal puede generalizar.

**Crear un Modelo Polinómico de Grado 5**

Usa `Pipeline` para combinar `PolynomialFeatures` (grado 5), `StandardScaler` y `LinearRegression`. Entrénalo con los datos de entrenamiento.

```python
# Crear el pipeline para el modelo polinómico
### TU CÓDIGO AQUÍ ###
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LinearRegression

modelo_polinomico = Pipeline([
    ('poly', PolynomialFeatures(degree=5)),
    ('scaler', StandardScaler()),
    ('regresion', LinearRegression())
])
```

```python
# Entrenar el modelo
### TU CÓDIGO AQUÍ ###
modelo_polinomico.fit(X_train, y_train)
```

```python
# Calcular el error (RMSE) en entrenamiento y prueba y realizar un print de estos
### TU CÓDIGO AQUÍ ###
from sklearn.metrics import mean_squared_error

y_train_pred = modelo_polinomico.predict(X_train)
y_test_pred = modelo_polinomico.predict(X_test)

rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

print("RMSE de entrenamiento:", rmse_train)
print("RMSE de prueba:", rmse_test)

rmse_train_polinomico = rmse_train
rmse_test_polinomico = rmse_test
```

**Pregunta:** ¿Qué indica la diferencia que observas entre el error de entrenamiento y el de prueba? ¿Le recomendarías este modelo a la firma inmobiliaria? ¿Por qué?

El RMSE de entrenamiento es de 34.152,35 mientras que el de prueba es de 41.101,60. Esto significa que el modelo comete menos errores con los datos que utilizó para aprender, pero tiene más dificultades cuando debe predecir precios de viviendas nuevas. Por eso no lo recomendaría como primera opción, ya que necesitamos un modelo que no solo aprenda los datos que conoce sino que también pueda hacer buenas predicciones con información nueva.

### **Ejercicio 4: Aplicar Regularización**

Ahora, vamos a "curar" el sobreajuste. Usaremos los mismos `PolynomialFeatures` de grado 5, pero cambiaremos el modelo de regresión.

**Implementar Regresión Ridge**

Copia el pipeline anterior, pero reemplaza `LinearRegression` con `Ridge(alpha=10)`. `alpha` es la fuerza de la penalización.

```python
# Crear el pipeline para Ridge
### TU CÓDIGO AQUÍ ###
from sklearn.linear_model import Ridge

modelo_ridge = Pipeline([
    ('poly', PolynomialFeatures(degree=5)),
    ('scaler', StandardScaler()),
    ('regresion', Ridge(alpha=10))
])
```

```python
# Entrenar el modelo
### TU CÓDIGO AQUÍ ###
modelo_ridge.fit(X_train, y_train)
```

```python
# Calcular el error (RMSE) en entrenamiento y prueba y realizar un print de estos
### TU CÓDIGO AQUÍ ###
y_train_pred = modelo_ridge.predict(X_train)
y_test_pred = modelo_ridge.predict(X_test)

rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

print("RMSE de entrenamiento:", rmse_train)
print("RMSE de prueba:", rmse_test)

rmse_train_ridge = rmse_train
rmse_test_ridge = rmse_test
```

**Interpreta los resultados:** El RMSE de entrenamiento es de 35.204,89 y el de prueba es de 35.206,63. Como ambos valores son muy parecidos el modelo tiene un comportamiento similar con los datos de entrenamiento y los de prueba. Esto indica que logra generalizar bien y que no presenta una diferencia importante entre ambos conjuntos.

**Implementar Regresión Lasso**

Haz lo mismo, pero ahora con `Lasso(alpha=500, max_iter=10000)`. Lasso necesita un `alpha` más grande (porque la penalización L1 es diferente) y a veces más `max_iter` para converger.

```python
# Crear el pipeline para Lasso
### TU CÓDIGO AQUÍ ###
from sklearn.linear_model import Lasso

modelo_lasso = Pipeline([
    ('poly', PolynomialFeatures(degree=5)),
    ('scaler', StandardScaler()),
    ('regresion', Lasso(alpha=500, max_iter=10000))
])
```

```python
# Entrenar el modelo
### TU CÓDIGO AQUÍ ###
modelo_lasso.fit(X_train, y_train)
```

```python
# Calcular el error (RMSE) en entrenamiento y prueba y realizar un print de estos
### TU CÓDIGO AQUÍ ###
y_train_pred = modelo_lasso.predict(X_train)
y_test_pred = modelo_lasso.predict(X_test)

rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

print("RMSE de entrenamiento:", rmse_train)
print("RMSE de prueba:", rmse_test)

rmse_train_lasso = rmse_train
rmse_test_lasso = rmse_test
```

**Interpreta los resultados:**  El RMSE de entrenamiento es de 35.352,92 y el de prueba es de 35.447,71. La diferencia entre ambos errores es pequeña lo que indica que el modelo también funciona de manera similar con los datos conocidos y los nuevos. Sin embargo su error de prueba es ligeramente mayor que el de Ridge, por lo que sus predicciones son un poco menos precisas. Aun así Lasso tiene la ventaja de eliminar algunas variables que no aportan al modelo.

**Selección de Variables con Lasso**

Una de las grandes ventajas de Lasso es que puede eliminar variables. Vamos a ver cuántas de las características polinómicas que creamos (ej. `OverallQual^2`, `GrLivArea^5`, `OverallQual * GrLivArea^4`, etc.) fueron eliminadas.

```python
# Extraer los coeficientes del modelo Lasso
### TU CÓDIGO AQUÍ ###
coeficientes = modelo_lasso.named_steps['regresion'].coef_
```

```python
# Extraer los nombres de las características generadas
# Guíate por el siguiente código: feature_names = pipeline_lasso.named_steps['poly_features'].get_feature_names_out(X.columns)
### TU CÓDIGO AQUÍ ###
feature_names = modelo_lasso.named_steps['poly'].get_feature_names_out(X.columns)
```

```python
# Contar cuántos coeficientes son exactamente cero
### TU CÓDIGO AQUÍ ###
n_eliminadas = np.sum(coeficientes == 0)
```

```python
# Realizar un print de:
# Número total de características polinómicas generadas
# Número de características eliminadas por Lasso (coef = 0)
# Número de características CONSERVADAS
# Porcentaje de variables eliminadas
# Lista con los nombres de las características conservadas por Lasso (coef != 0)

### TU CÓDIGO AQUÍ ###
# Realizar un print de los resultados
n_total = len(feature_names)
n_eliminadas = np.sum(coeficientes == 0)
n_conservadas = np.sum(coeficientes != 0)
porcentaje_eliminadas = (n_eliminadas / n_total) * 100

print("Número total de características polinómicas generadas:", n_total)
print("Número de características eliminadas por Lasso (coef = 0):", n_eliminadas)
print("Número de características conservadas (coef != 0):", n_conservadas)
print("Porcentaje de variables eliminadas:", round(porcentaje_eliminadas, 2), "%")

print("\nCaracterísticas conservadas por Lasso:")
for nombre, coef in zip(feature_names, coeficientes):
    if coef != 0:
        print(nombre)
```

**Interpreta los resultados:**  Lasso generó 21 características polinómicas y eliminó 15, lo que equivale al 71,43 % del total. Conservó seis características: OverallQual, OverallQual GrLivArea, OverallQual^2 GrLivArea, OverallQual^3 GrLivArea, OverallQual^5 y GrLivArea^5. Esto muestra que Lasso puede reducir la cantidad de variables que utiliza el modelo, dejando aquellas cuyos coeficientes son diferentes de cero. De esta manera facilita la interpretación de las predicciones y permite trabajar con un modelo menos complejo.
