### **Ejercicio 19: Márgenes por Categoría**
Dadas las listas `categorias = ["Bebidas", "Aseo", "Snacks"]` y `margen_pct = [25, 18, 32]`, que contienen el margen de utilidad de cada categoría en porcentaje. Crea un diccionario `margenes` uniendo ambas listas.

```python
### TU CÓDIGO AQUÍ ###
categorias = ["Bebidas", "Aseo", "Snacks"]
margen_pct = [25, 18, 32]

margenes = dict(zip(categorias, margen_pct))

print(margenes)
```

### **Ejercicio 20: Margen Total**
Recorre el diccionario `margenes` con un ciclo `for` y suma todos los márgenes en una variable llamada `suma_margenes`. Recuerda crear la variable en `0` antes del ciclo y acumular con `+=`.

```python
### TU CÓDIGO AQUÍ ###
suma_margenes = 0

for margen in margenes.values():
    suma_margenes += margen

print(f"La suma de los márgenes es: {suma_margenes}")
```

```python
import hashlib

_SAL = "extdo-retail-2026-II"


def _canon(x):
    if isinstance(x, bool):
        return repr(x)
    if isinstance(x, dict):
        return "D[" + ",".join(_canon(k) + ":" + _canon(v) for k, v in sorted(x.items(), key=str)) + "]"
    if isinstance(x, (list, tuple)):
        return "L[" + ",".join(_canon(i) for i in x) + "]"
    if isinstance(x, float) and x == int(x):
        return str(int(x))
    return repr(x)


def _huella(n, valor):
    return hashlib.sha256((_SAL + "|" + str(n) + "|" + _canon(valor)).encode("utf-8")).hexdigest()[:16]


def verificar_taller():
    pruebas = [
        ( 1, "b5e9fc903a52ebfa", lambda: producto_1["nombre"]),
        ( 2, "82cddf1fbb43027c", lambda: precio_actual),
        ( 3, "ba158be199abd88f", lambda: producto_1["precio"]),
        ( 4, "67f26d920a7daa6c", lambda: es_top_ventas),
        ( 5, "0022208bcc07108b", lambda: aplica_promocion),
        ( 6, "144963941a5e8637", lambda: producto_1["categoria"]),
        ( 7, "ee01b5c599154165", lambda: [unidades_retiradas, "unidades_vendidas" in producto_1]),
        ( 8, "0b440a60b760ba38", lambda: falta_descuento),
        ( 9, "7213992afecd73ae", lambda: descuento),
        (10, "0bbcebb192006c3c", lambda: [producto_1.get("canal"), producto_1.get("stock"), len(producto_1)]),
        (11, "b3ef303b1d0c98b6", lambda: ventas_ciudades),
        (12, "6380f0e16ede71b1", lambda: venta_cali_marzo),
        (13, "a7b76736c5590fc4", lambda: sorted(ciudades, key=str)),
        (14, "ad03f2042e22e28a", lambda: n_ciudades),
        (15, "38783cd4628a7326", lambda: meses),
        (16, "4bcf21a15f6c1506", lambda: dias_promo),
        (17, "ed61a6cf65367803", lambda: tienda_online),
        (18, "0c6599bc7e9f6331", lambda: costo_internacional),
        (19, "57edafd7a757c604", lambda: margenes),
        (20, "951f1d8d80a441e4", lambda: suma_margenes),
    ]

    resultados = {}
    for numero, esperado, obtener in pruebas:
        try:
            resultados[numero] = (_huella(numero, obtener()) == esperado)
        except Exception:
            resultados[numero] = False

    correctos = 0
    for numero in resultados:
        if resultados[numero]:
            correctos += 1
    total = len(pruebas)
    nota = round(correctos * 20 / total, 1)

    print("=" * 60)
    print("  VERIFICACIÓN AUTOMÁTICA DEL TALLER")
    print("=" * 60)

    fila = ""
    for numero, _esperado, _obtener in pruebas:
        icono = "✅" if resultados[numero] else "❌"
        fila += "  Ej %2d: %s " % (numero, icono)
        if numero % 5 == 0:
            print(fila)
            fila = ""
    if fila:
        print(fila)

    print("-" * 60)
    print("  RESULTADO: %d de %d ejercicios correctos" % (correctos, total))
    print("  PUNTAJE:   %s de 20 puntos" % nota)
    print("=" * 60)

    pendientes = []
    for numero, _esperado, _obtener in pruebas:
        if not resultados[numero]:
            pendientes.append(str(numero))

    if not pendientes:
        print("  🎉 ¡Felicidades! El reporte de analítica está impecable.")
    else:
        print("  Revisa estos ejercicios: " + ", ".join(pendientes))
        print("  Pista: verifica el nombre exacto de la variable, las llaves del")
        print("  diccionario y los límites del range.")


verificar_taller()
```
